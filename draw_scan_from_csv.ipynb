```python
theta = angle + np.pi / 2
r = arr[idx,:]
x = r * np.cos(theta)
y = r * np.sin(theta)
# idx += 1
plt.plot(x, y)
plt.figure()
plt.plot(x, y, 'x')
```

```python
robot_width = 0.6
robot_half_width = robot_width / 2
```

```python
print(x.shape, y.shape)
mask_x = -robot_half_width <= x
mask_x &= x <= robot_half_width
filtered_x = x[mask_x]
filtered_y = y[mask_x]
print(filtered_x.shape, filtered_y.shape)
plt.plot(filtered_x, filtered_y, 'x')
```
